```python
# 2) Генерация траекторий
def generate_Morris_new(problem,phase_methods, N=6000, num_levels=100):
    param_values = morris_sampler.sample(problem, N, num_levels)
    i=0
    from tqdm import tqdm
    Y_mass = []
    Param_mass=[]
    for i in range(0, len(phase_methods), 1):
        Y = []
        P=[]

        # tqdm оборачивает param_values и показывает прогресс
        for j, p in enumerate(tqdm(param_values, desc=f"Метод {phase_methods[i].__name__}", unit=" шага")):
            F, phase_real, fs_mult, duration_T, gen_unstable, bits = p

            duration = duration_T / F
            Hz_unstable=gen_unstable*F*10e-6
            fs = F * fs_mult

            t = np.linspace(0, duration, int(duration * fs))

            phase = md.generate_common_phase(
                t,
                F,
                gen_unstable,
                phase0_deg=0
            )

            # 2. Сигналы с заданной фазовой разницей
            U, I = md.generate_signals_from_phase(
                phase,
                phase_diff_deg=phase_real
            )

            U=md.adc_quantize(U,int(bits),1)
            I=md.adc_quantize(I,int(bits),1)

            #print(p)

            phase_mes = phase_methods[i](t, U, I, F)
            phase_error = abs(phase_mes) - abs(phase_real)

            Y.append(abs(phase_error))
            P.append(p)

        Y_mass.append(np.array(Y))
        Param_mass.append(P)

    return Y_mass, Param_mass, param_values
```

```python
import os
import pickle
import hashlib
import numpy as np

def compute_hash(param_values):
    """Создаёт уникальный хеш для массива параметров"""
    return hashlib.md5(param_values.tobytes()).hexdigest()


def save_morris_data(filename, Y_mass, Param_mass, param_values):
    new_hash = compute_hash(param_values)

    # если файла нет — просто сохраняем
    if not os.path.exists(filename):
        with open(filename, "wb") as f:
            pickle.dump({
                "Y_mass": Y_mass,
                "Param_mass": Param_mass,
                "param_values": param_values,
                "hash": new_hash
            }, f)
        print("Новый файл")
        return

    # если файл есть — проверяем дубликаты
    existing_hashes = set()

    with open(filename, "rb") as f:
        while True:
            try:
                data = pickle.load(f)
                if "hash" in data:
                    existing_hashes.add(data["hash"])
            except EOFError:
                break

    if new_hash in existing_hashes:
        print("Сохранение пропущено")
        return

    # если не дубликат — сохраняем
    with open(filename, "ab") as f:
        pickle.dump({
            "Y_mass": Y_mass,
            "Param_mass": Param_mass,
            "param_values": param_values,
            "hash": new_hash
        }, f)

    print("Данные добавлены")
```

```python
def load_morris_data(filename):
    Y_mass_all = None
    Param_mass_all = None
    param_values_all = []

    if not os.path.exists(filename):
        return [], [], np.array([])

    with open(filename, "rb") as f:
        while True:
            try:
                data = pickle.load(f)

                if Y_mass_all is None:
                    Y_mass_all = data["Y_mass"]
                    Param_mass_all = data["Param_mass"]
                else:
                    for i in range(len(Y_mass_all)):
                        Y_mass_all[i] = np.concatenate([Y_mass_all[i], data["Y_mass"][i]])
                        Param_mass_all[i].extend(data["Param_mass"][i])

                param_values_all.append(data["param_values"])

            except EOFError:
                break

    param_values_all = np.vstack(param_values_all)

    return Y_mass_all, Param_mass_all, param_values_all
```

```python
import pandas as pd
import numpy as np
import openpyxl

# ---------- IEEE formatter ----------
def fmt_ieee(x, sig=3):
    if pd.isna(x):
        return ""

    if x == 0:
        return "0"

    exp = int(np.floor(np.log10(abs(x))))

    if -2 <= exp <= 2:
        return f"{x:.{sig-1}f}".rstrip('0').rstrip('.')

    mant = x / 10**exp
    mant_str = f"{mant:.{sig-1}f}".rstrip('0').rstrip('.')
    return f"{mant_str}e{exp}"

# ---------- top-3 bold ----------
def highlight_top3(row):
    # строка сейчас строковая → возвращаемся к float
    vals = row.astype(float)

    # берём модуль
    abs_vals = vals.abs()

    # индексы топ-3 по |μ|
    top3 = abs_vals.nlargest(3).index

    return [
        'font-weight: bold' if col in top3 else ''
        for col in row.index
    ]


# ---------- DataFrame ----------
df = pd.DataFrame(rows)

df_mu = df.pivot_table(
    index='Method',
    columns='Parameter',
    values='mu'
)

# порядок
df_mu = df_mu.reindex(columns=problem['names'])
df_mu = df_mu.reindex(index=[clean_name(m) for m in phase_methods])

# превращаем числа в строки ДО Excel
df_mu_str = df_mu.applymap(fmt_ieee)

styled = (
    df_mu_str
    .style
    .apply(highlight_top3, axis=1)
)

styled.to_excel(
    "morris_mu_highlighted.xlsx",
    engine="openpyxl"
)
```

```python
import numpy as np
import pickle

def clean_all_outliers(filename, problem, phase_methods, n_bins=100, threshold=3):
    # ===== загрузка =====
    with open(filename, "rb") as f:
        data = pickle.load(f)

    Y_mass_all = data["Y_mass"]
    Param_mass_all = data["Param_mass"]
    param_values_all = data["param_values"]

    bit_idx = problem['names'].index('Bit')

    total_fixed = 0

    # ===== по всем методам =====
    for m, method in enumerate(phase_methods):
        print(f"\n=== Метод: {method.__name__} ===")

        P = np.asarray(Param_mass_all[m])
        Y = np.asarray(Y_mass_all[m])
        X = P[:, bit_idx]

        # bins как в графике
        bins = np.linspace(X.min(), X.max(), n_bins + 1)
        bin_indices = np.digitize(X, bins) - 1

        # средние по bin
        Y_bin = np.array([
            Y[bin_indices == i].mean() if np.any(bin_indices == i) else np.nan
            for i in range(n_bins)
        ])

        # лог шкала
        logY = np.log10(Y_bin)

        # медиана и отклонение
        median = np.nanmedian(logY)
        diff = np.abs(logY - median)

        # выбросы
        outlier_bins = np.where(diff > threshold)[0]

        print(f"Найдено выбросов (bins): {len(outlier_bins)}")

        for b in outlier_bins:
            mask = (bin_indices == b)
            indices_in_bin = np.where(mask)[0]

            if len(indices_in_bin) == 0:
                continue

            # худшая точка в bin
            Y_candidates = Y[indices_in_bin]
            local_idx = np.argmax(Y_candidates)
            global_idx = indices_in_bin[local_idx]

            # соседний bin
            neighbor_bin1 = b + 1 #if b < n_bins - 1 else b - 1
            neighbor_bin2 = b - 1
            neighbor_bin=(neighbor_bin1+neighbor_bin2)/2
            mask_neighbor = (bin_indices == neighbor_bin)
            neighbor_values = Y[mask_neighbor]

            if len(neighbor_values) == 0:
                continue

            replacement = np.median(neighbor_values)

            print(f"\n🔧 Исправление:")
            print(f"  Bin {b} ({bins[b]:.2f}-{bins[b+1]:.2f})")
            print(f"  Индекс: {global_idx}")
            print(f"  Было: {Y[global_idx]}")
            print(f"  Стало: {replacement}")

            # замена
            Y_mass_all[m][global_idx] = replacement
            total_fixed += 1

    # ===== сохранение =====
    with open(filename, "wb") as f:
        pickle.dump({
            "Y_mass": Y_mass_all,
            "Param_mass": Param_mass_all,
            "param_values": param_values_all
        }, f)

    print(f"\n✅ Готово! Исправлено точек: {total_fixed}")
```

```python
def binned_mean(X, Y, bins):
    bin_idx = np.digitize(X, bins)

    X_bin = []
    Y_bin = []

    for b in range(1, len(bins)):
        mask = bin_idx == b
        if np.any(mask):
            X_bin.append(X[mask].mean())
            Y_bin.append(Y[mask].mean())

    return np.array(X_bin), np.array(Y_bin)
```

```python
from scipy.optimize import curve_fit
import numpy as np

def exp_model(x, a, b, c):
    return a * np.exp(-b * x) + c


def fit_exponential(x, y):
    try:
        # начальные приближения
        a0 = y.max() - y.min()
        b0 = 1 / (x.max() - x.min() + 1e-9)
        c0 = y.min()

        popt, _ = curve_fit(exp_model, x, y, p0=(a0, b0, c0), maxfev=10000)

        y_fit = exp_model(x, *popt)

        # R^2
        ss_res = np.sum((y - y_fit)**2)
        ss_tot = np.sum((y - np.mean(y))**2)
        r2 = 1 - ss_res / ss_tot

        return popt, r2
    except:
        return None, None
```

```python
# ===== настройки =====
Param_index = problem['names'].index('FsM')   # ['F', 'φ', 'FsM','DT','GU','Bit']
param_name = problem['names'][Param_index]

n_bins = 25
colors = plt.cm.tab10(np.linspace(0, 1, len(phase_methods)))

plt.figure(figsize=(6, 4))

# ===== общий диапазон bins =====
all_X = []
for m in range(len(phase_methods)):
    P = np.asarray(Param_mass_all[m])
    all_X.append(P[:, Param_index])

all_X = np.concatenate(all_X)
bins = np.linspace(all_X.min(), all_X.max(), n_bins + 1)

saturation_points = {}
fit_quality = {}

for m, method in enumerate(phase_methods):
    P = np.asarray(Param_mass_all[m])
    Y = np.asarray(Y_mass_all[m])

    X_raw = P[:, Param_index]
    Y_raw = Y

    # binning
    X_bin, Y_bin = binned_mean(X_raw, Y_raw, bins)

    # удаляем nan
    mask = ~np.isnan(X_bin) & ~np.isnan(Y_bin)
    X_bin = X_bin[mask]
    Y_bin = Y_bin[mask]

    # ===== фит экспоненты =====
    popt, r2 = fit_exponential(X_bin, Y_bin)

    if popt is not None and r2 > 0.9 and popt[1] > 0:
        a, b, c = popt

        # плато: близость к асимптоте
        threshold = c + 0.05 * (Y_bin.max() - c)

        idx = np.where(Y_bin <= threshold)[0]

        if len(idx) > 0:
            x_plateau = X_bin[idx[0]]
            saturation_points[method.__name__] = x_plateau
            fit_quality[method.__name__] = r2

            #plt.axvline(x_plateau, color=colors[m], linestyle='--', alpha=0.4)

            # рисуем фит
            x_dense = np.linspace(X_bin.min(), X_bin.max(), 200)
            y_fit = exp_model(x_dense, *popt)

            #plt.plot(x_dense, y_fit, '--', color=colors[m], alpha=0.6)

        else:
            saturation_points[method.__name__] = None
            fit_quality[method.__name__] = r2
    else:
        # экспонента не подходит
        saturation_points[method.__name__] = None
        fit_quality[method.__name__] = r2

    # исходный график
    plt.plot(
        X_bin,
        Y_bin,
        color=colors[m],
        #marker='o',
        #linestyle='',
        label=f"{clean_name(method)}"
    )
    

# ===== оформление =====
plt.yscale('log')
plt.xlabel(param_name)
plt.ylabel('Mean phase error')
plt.grid(True, alpha=0.3)
plt.legend(fontsize=9)
plt.tight_layout()
plt.show()
```
